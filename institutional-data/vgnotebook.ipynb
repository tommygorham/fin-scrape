```python
import pandas as pd

# Load the csv (assuming the filename based on your screenshot)
df = pd.read_csv('vanguard_holdings_cleaned.csv')

# Display the first few rows to confirm
df.head()
```

```python
# Create a temporary 'Impact' column
df['Impact'] = df['% of Portfolio'] * df['% Change']

# Sort by "impact" to see the biggest contributors to performance
impact_df = df.sort_values(by='Impact', ascending=False)

impact_df[['Stock', '% of Portfolio', '% Change', 'Impact']].head(25)
```

```python
# 1. Filter: Keep only rows where Portfolio % is greater than 1.0
significant_holdings = df[df['% of Portfolio'] > 1.0]

# 2. Sort: Order by % Change (Highest to Lowest)
sorted_df = significant_holdings.sort_values(by='% Change', ascending=False)

# Display the result
sorted_df
```
